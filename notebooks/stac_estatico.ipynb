# STAC estático

Catalog, Collection e Item describen un activo ráster local.

```python
from pathlib import Path
import json
import os

CWD = Path.cwd()
ROOT = CWD if (CWD / 'data').is_dir() else CWD.parent
MODE = os.environ.get('COURSE_DATA_MODE', 'fixtures')
assert MODE in ('fixtures', 'local')
```

```python
base = ROOT / 'data/fixtures/stac'
catalog = json.loads((base / 'catalog.json').read_text())
collection = json.loads((base / 'collection.json').read_text())
item = json.loads((base / 'item-referencia.json').read_text())
asset = (base / item['assets']['data']['href']).resolve()
assert catalog['type'] == 'Catalog'
assert collection['id'] == 'referencia'
assert asset.is_file()
item['assets']['data']['title']
```
