# WFS clásico con fixture local

Este notebook no consulta Internet.

```python
from pathlib import Path
import json
import os

CWD = Path.cwd()
ROOT = CWD if (CWD / 'data').is_dir() else CWD.parent
MODE = os.environ.get('COURSE_DATA_MODE', 'fixtures')
assert MODE in ('fixtures', 'local')
CONFIG = json.loads((ROOT / 'notebooks/config.local.json').read_text()) if MODE == 'local' else {}
```

```python
import xml.etree.ElementTree as ET
if MODE == 'local':
    import requests
    response = requests.get(CONFIG['wfs_capabilities'], timeout=30)
    response.raise_for_status()
    root = ET.fromstring(response.content)
else:
    xml = ROOT / 'data/fixtures/responses/ogc-clasico/wfs-capabilities.xml'
    root = ET.parse(xml).getroot()
ns = {'wfs': 'http://www.opengis.net/wfs/2.0'}
name = root.findtext('.//wfs:FeatureType/wfs:Name', namespaces=ns)
assert name == 'curso:referencia'
name
```
