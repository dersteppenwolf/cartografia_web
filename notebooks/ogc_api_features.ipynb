# OGC API - Features con fixtures locales

Compara la colección con el nombre WFS.

```python
from pathlib import Path
import json
import os

CWD = Path.cwd()
ROOT = CWD if (CWD / 'data').is_dir() else CWD.parent
MODE = os.environ.get('COURSE_DATA_MODE', 'fixtures')
assert MODE in ('fixtures', 'local')
CONFIG = json.loads((ROOT / 'notebooks/config.local.json').read_text()) if MODE == 'local' else {}
```

```python
if MODE == 'local':
    import requests
    landing_url = CONFIG['ogc_api_landing']
    landing = requests.get(landing_url, timeout=30).json()
    conformance = requests.get(f'{landing_url}/conformance', timeout=30).json()
    collections = requests.get(f'{landing_url}/collections', timeout=30).json()
    collection_id = collections['collections'][0]['id']
    items = requests.get(f'{landing_url}/collections/{collection_id}/items', timeout=30).json()
else:
    base = ROOT / 'data/fixtures/responses/ogc-api-features'
    landing = json.loads((base / 'landing.json').read_text())
    conformance = json.loads((base / 'conformance.json').read_text())
    collections = json.loads((base / 'collections.json').read_text())
    items = json.loads((base / 'items.json').read_text())
assert collections['collections'][0]['id'].endswith('referencia')
assert any(feature['id'].endswith('zona-1') for feature in items['features'])
assert any('ogcapi-features' in value for value in conformance['conformsTo'])
collections['collections'][0]['title']
```
